```python
#7.5 capitalize the element in things that revers to a person and print the list. does the element change?
for thing in things:
    if thing == 'cinderella':
        things[1] = thing.title()
        print(things) # Yes, it changed from cinderella to Cinderella.
```

```python
#9.2 define a generator function called get_odds() that returns the odd numbers from range(10). use a for loop to find and print the third value 
def get_odds():
    'function that looks through range for odd numbers'
    for number in range(10):
        if number % 2 != 0:
            yield number
# for loop to print 3rd value
count = 0
for value in get_odds():
    count += 1
    if count == 3:
        print(value)
        break
```
