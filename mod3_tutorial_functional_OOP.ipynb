```python
# sort 0s, 1s, and 2s
class Solution:
    def sort012(self, arr):
        # code here
        n = len(arr)
        lo = 0
        hi = n - 1
        mid = 0
        # dutch national flag algorithm
        while mid <= hi:
             if arr[mid] == 0:
                arr[lo], arr[mid] = arr[mid], arr[lo]
                lo = lo + 1
                mid = mid + 1
             elif arr[mid] == 1:
                mid = mid + 1
             else:
                arr[mid], arr[hi] = arr[hi], arr[mid]
                hi = hi - 1
        return arr
    # did have to use the editorial function for help
```

```python
# First Occurence in Sorted 
class Solution:
    def firstSearch(self, arr, k):
        # Code Here
        low = 0
        high = len(arr) - 1
        ans = - 1
        # modified binary search
        while low <= high:
            mid = low + (high-low) // 2

            if arr[mid] == k:
                ans = mid
                high = mid -1
            elif arr[mid] < k:
                low = mid +1
            else:
                high = mid - 1
        return ans
    # did use https://www.geeksforgeeks.org/dsa/find-first-occurrence-in-sorted/ for help
```
