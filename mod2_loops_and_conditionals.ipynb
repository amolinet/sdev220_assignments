```python
#4.1 assign values to secret and guess, use if, elif, else to print a string depending on certain conditions.
secret = 3
guess = 6
if guess > secret:
    print('Too high')
elif guess < secret:
    print('Too low')
else:
    print('Just right')
    
```

```python
#4.2 assign small and green to true or false, use if/elif/else to determine if it matches pea, cherry, watermelon, or pumpkin. 
small = "true"
green = "false" 
if green == "true" and small == "true":
    print("It's a pea!")
elif green == "true" and small == "false":
    print("It's a watermelon!")
elif green == "false" and small == "true":
    print("It's a cherry!")
else:
    print("It's a pumpkin!")
```

```python
#6.2 guess me game
guess_me = 7
number = 1
while True:
    if number < guess_me:
        print('too low')
    elif number == guess_me:
        print('Just right!')
        break
    else:
        print('oops')
        break
    number += 1
```

```python
#6.3 guess_me with for loop
guess_me = 5
for number in range(10):
    if number < guess_me:
        print('too low')
    elif number == guess_me:
        print('found it!')
        break
    else:
        print('oops')
        break
number += 1
```
