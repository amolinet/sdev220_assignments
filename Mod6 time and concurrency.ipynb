```python
# 13.1 Write the current date as a string to the text file today.txt.

from datetime import datetime, timedelta
from zoneinfo import ZoneInfo

cst = ZoneInfo('America/Chicago')
fmt = '%m/%d/%Y'
current_date = datetime.now(cst)
current_date_str = current_date.strftime(fmt)

with open('today.txt', 'w') as file:
    file.write(current_date_str)

print(current_date_str)
```

```python
#13.2 Read the text file today.txt into the string today_string.
with open ('today.txt', 'r') as file:
    today_string = file.read()
    print(today_string)
```

```python
#13.3 parse the date string
today_string = datetime.strptime(today_string, fmt)
print(today_string)
```

```python
import multiprocess
import random
import time

def process():
    wait_time = random.random()
    time.sleep(wait_time)
    current_time = time.strftime('Current Time: %H:%M:%S')
    print(current_time)

if __name__ == '__main__':
    processes = []
    for _ in range(3):
        p = multiprocessing.Process(target=process)
        processes.append(p)
        p.start()

    for p in processes:
        p.join()
```
