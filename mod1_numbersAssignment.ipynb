```python
#3.1 and 3.2 how many seconds are in an hour, assign result to variable seconds_per_hour
seconds_per_minute = 60
minutes_per_hour = 60
seconds_per_hour = seconds_per_minute * minutes_per_hour
print("seconds per hour: ", seconds_per_hour)
```

```python
#3.3 and 3.4 how many seconds per day, use the seconds_per_hour variable. Assign result to seconds_per_day
hours_per_day = 24
seconds_per_day = hours_per_day * seconds_per_hour
print("seconds per day: ", seconds_per_day)
```

```python
#3.5 use floating point division to divide seconds_per_day by seconds_per_hour
fpResult = seconds_per_day / seconds_per_hour
print("result: ", fpResult)
```

```python
#3.6 use integer division to divide seconds_per_day by seconds_per_hour, did this number agree with 3.5 other than the .0?
intResult = seconds_per_day // seconds_per_hour
print("result: ", intResult) #yes, other than the .0, the results agree.
```
